```python
import uuid
import random
import time
from kafka import KafkaProducer
import json

producer = KafkaProducer(
    bootstrap_servers='kafka-3ad4f62-project-746a.c.aivencloud.com:24264',
    security_protocol="SSL",
    ssl_cafile="/content/ca.pem",
    ssl_certfile="/content/service.cert",
    ssl_keyfile="/content/service.key",
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

topic = "ride_requests"

locations = ["San José", "Cartago", "Heredia", "Alajuela"]

def generate_event():
    return {
        "ride_id": str(uuid.uuid4()),
        "user_id": random.randint(1,100),
        "pickup": random.choice(locations),
        "destination": random.choice(locations)
    }

i = 0
while i < 10:
    event = generate_event()

    producer.send(topic, event)
    print("New ride request:", event)

    time.sleep(1)

    i += 1
```
